# Global airport network

This example downloads Natural Earth's global airport dataset and renders airports as a pickable Lonboard `ScatterplotLayer`. Airport type controls the color, while the GPU-rendered points stay visible across zoom levels. Hover over an airport to inspect its name and type.

```python
# /// script
# dependencies = [
#   "geopandas",
#   "lonboard",
#   "matplotlib",
#   "pyarrow",
# ]
# ///

import geopandas as gpd
import matplotlib.colors as mcolors
import matplotlib.pyplot as plt
from lonboard import Map, ScatterplotLayer

url = "https://d2ad6b4ur7yvpq.cloudfront.net/naturalearth-3.3.0/ne_10m_airports.geojson"
airports = gpd.read_file(url)

airport_types = airports["type"].astype("category").cat.codes.to_numpy()
norm = mcolors.Normalize(vmin=airport_types.min(), vmax=airport_types.max())
colors = (plt.colormaps["viridis"](norm(airport_types)) * 255).astype("uint8")

layer = ScatterplotLayer.from_geopandas(
    airports[["geometry", "name", "type"]],
    get_fill_color=colors,
    get_radius=15000,
    radius_min_pixels=2,
    radius_max_pixels=15,
    pickable=True,
)

Map(
    layers=[layer],
    view_state={
        "longitude": 0.0,
        "latitude": 20.0,
        "zoom": 1.5,
        "pitch": 0,
        "bearing": 0,
    },
    height=700,
    show_tooltip=True,
)
```
